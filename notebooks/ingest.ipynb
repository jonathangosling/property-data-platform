```python
import os
import shutil
import sys

# Resolve the bundle files root from the notebook's workspace path.
notebook_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
files_root = '/Workspace' + '/'.join(notebook_path.split('/')[:-2])
src_workspace = os.path.join(files_root, 'src')

# Copy src/ to /tmp — workspace filesystem has I/O issues when scanned for imports.
src_local = '/tmp/property_data_platform_src'
if os.path.exists(src_local):
    shutil.rmtree(src_local)
shutil.copytree(src_workspace, src_local)
sys.path.insert(0, src_local)
```
